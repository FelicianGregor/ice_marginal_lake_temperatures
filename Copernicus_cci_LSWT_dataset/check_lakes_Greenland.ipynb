## check lakes in greenland
This is a short script to check whoich lake id the lake Kangaarsuup Tasersua has in the Copernicus lake dataset `lakescci_v210`. There is a csv metadata file with all the centroids of the lakes, so lets check this.



```python
# import
import geopandas as gpd
import os
import pathlib
import pandas as pd
```

```python
lakes_df.columns
```

```python
# read csv file
lakes_df = pd.read_csv('lakescci_v2.1.0_metadata.csv', sep = ";")

# 2. Convert to a GeoDataFrame by generating Point geometries
lakes_gdf = gpd.GeoDataFrame(
    lakes_df, 
    geometry=gpd.points_from_xy(lakes_df['lon centre'], lakes_df['lat centre']),
    crs="EPSG:4326"  # EPSG:4326 is standard WGS84 (lat/lon)
)
```

```python
# filter for country = Greenland

greenland_lakes_gdf = lakes_gdf[lakes_gdf['country'] == 'Greenland']
greenland_lakes_gdf

```

```python
# interactive map to explore points and lakes
greenland_lakes_gdf.explore(
    color="blue",            
    marker_kwds=dict(radius=5, fill=True), 
    tiles="OpenStreetMap")
# great, the lake I am looking for is this one:
# id: 300003969 
```
